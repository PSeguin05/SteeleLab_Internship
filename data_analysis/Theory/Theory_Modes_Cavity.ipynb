# Eigenmodes of the cavity (Theory)

## Imports

```python
import numpy as np
import matplotlib.pyplot as plt

from cavity_theory import cavity_modes
```

## Plots

```python
thmodes = cavity_modes(radius = 25, # mm
                       lengths = np.linspace(63, 113, 200), # mm
                       freq_window = (4, 8) # GHz
                       )

thmodes.compute_modes()
thmodes.crossing_modes()
thmodes.non_crossing_modes()
```

```python
plt.figure(figsize=(9, 4))
thmodes.plot_modes(crossing=None, inverted_axis = True)
plt.title('Theoretical frequencies of TEM modes')
plt.ylabel('Cavity length [mm]')
plt.xlabel('Frequency [GHz]')
plt.tight_layout()
plt.savefig('Cavity_TEM.png')
plt.show()
```

```python
plt.figure(figsize=(6, 7))
thmodes.plot_modes(crossing=None, inverted_axis = False)
plt.title('Theoretical frequencies of TEM modes')
plt.xlabel('Cavity length [mm]')
plt.ylabel('Frequency [GHz]')
plt.tight_layout()
plt.savefig('Cavity_TEM.png')
plt.show()
```
