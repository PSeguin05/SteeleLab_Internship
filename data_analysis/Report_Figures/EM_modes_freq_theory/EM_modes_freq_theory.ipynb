```python
import numpy as np
import matplotlib.pyplot as plt

from microwave_cavity import cavity_theory
```

```python
thmodes = cavity_theory(radius = 25, # mm
                       lengths = np.linspace(60, 120, 200), # mm
                       freq_window = (4, 8) # GHz
                       )

thmodes.compute_modes()
```

```python
plt.figure(figsize=(6, 6))
thmodes.plot_modes(crossing=None)
plt.title('Cavity transversal modes')
plt.xlabel('Cavity length [mm]')
plt.ylabel('Frequency [GHz]')
plt.grid(True)
plt.show()
```
