## Functions

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.colors import LogNorm
from matplotlib.colors import SymLogNorm
```

```python
def plot_cavity_eigenmodes(filepath: str):
    """
    Plot eigenmodes from the simulation.

    Args:
        filepath (str): Path to the data

    Returns:
        None
    """
    file = pd.read_csv(filepath)
    n = file.shape[1]
    
    lengths = file.iloc[:, 0].values
    freqs = file.iloc[:, 1:n].values
    
    plt.figure(figsize=(10,6))

    plt.plot(freqs*1e-9, lengths)
    
    plt.ylabel("Length (mm)")
    plt.xlabel("Frequency (GHz)")
    plt.title("Eigenmodes of the cavity")
    plt.grid()

    plt.tight_layout()
    plt.show()
```

```python
def plot_cavity_eigenQ(filepath: str):
    """
    Plot Q-factor of eigenmodes from the simulation.

    Args:
        filepath (str): Path to the data

    Returns:
        None
    """
    file = pd.read_csv(filepath)
    n = file.shape[1]
    
    lengths = file.iloc[:, 0].values
    Qs = file.iloc[:, 1:n].values
    
    plt.figure(figsize=(10,6))

    plt.plot(lengths, Qs)
    
    plt.xlabel("Length (mm)")
    plt.ylabel("Frequency (GHz)")
    plt.title("Q-factor of eigenmodes of the cavity")
    plt.grid()

    plt.tight_layout()
    plt.show()
```

```python
def plot_cavity_eigenmodes_colorbar_Q(filepath: str):
    """
    Plot eigenmodes from the simulation, with a colorbar for the quality factor

    Args:
        filepath (str): Path to the data

    Returns:
        None
    """
    file = pd.read_csv(filepath)
    n = file.shape[1]
    
    lengths = file.iloc[:, 0].values
    freqs = file.iloc[:, 1:(n//2+1)].values
    Qs = file.iloc[:, (n//2+1):n].values
    
    n_rows = len(lengths)
    n_modes = n//2
    
    x_freq, y_length, color_Q = [], [], []
    
    for i in range(n_rows):
        for j in range(n_modes):
            x_freq.append(freqs[i,j])
            y_length.append(lengths[i])
            color_Q.append(Qs[i,j])
    
    plt.figure(figsize=(10,7))
    
    scatter = plt.scatter(x_freq, y_length, c=color_Q, cmap='viridis', s=4)
    cbar = plt.colorbar(scatter)
    cbar.set_label('Quality factor Q')
    
    plt.ylabel("Length (mm)")
    plt.xlabel("Frequency (GHz)")
    plt.title("Eigenmodes of the cavity")
    plt.grid()

    plt.tight_layout()
    plt.show()
```

```python
def plot_cavity_eigenmodes_pin_length(filepath):
    """
    Plot eigenmodes from the simulation wrt the pin length inside the cavity.

    Args:
        filepath (str): Path to the data

    Returns:
        None
    """
    file = pd.read_csv(filepath)
    n = file.shape[1]
    
    pin_lengths = file.iloc[:, 0].values
    freqs = file.iloc[:, 1:n].values
    
    plt.figure(figsize=(10,6))
    
    plt.plot(- pin_lengths - 5, freqs*1e-9)

    
    plt.xlabel("Pin Length (mm)")
    plt.ylabel("Frequency (GHz)")
    plt.title("Eigenmodes of the cavity with a pin")
    plt.grid()

    plt.tight_layout()
    plt.show()
```

```python
def plot_cavity_eigenmodes_eigenQ_pin_length(filepath: str):
    """
    Plot eigenmodes and Q-factors from the simulation wrt the pin length inside the cavity.

    Args:
        filepath (str): Path to the data

    Returns:
        None
    """
    file = pd.read_csv(filepath)
    n = file.shape[1]
    
    pin_lengths = file.iloc[:, 0].values
    freqs = file.iloc[:, 1:(n//2+1)].values
    Qs = file.iloc[:, (n//2+1):n].values
    
    plt.figure(figsize=(10,6))
    ax_freq = plt.subplot(2, 1, 1)
    ax_Q = plt.subplot(2, 1, 2)
    
    ax_freq.plot(-pin_lengths - 5, freqs*1e-9)
    ax_Q.plot(-pin_lengths - 5, Qs)
    
    ax_Q.set_xlabel("Pin Length (mm)")
    ax_freq.set_ylabel("Frequency (GHz)")
    ax_Q.set_ylabel("Quality factor Q")
    ax_freq.grid()
    ax_Q.grid()
    
    plt.suptitle("Eigenmodes of the cavity with a pin")
    
    plt.tight_layout()
    plt.show()
```

```python
def plot_cavity_eigenmodes_eigenQ_pin_depth(filepath: str):
    """
    Plot eigenmodes and Q-factors from the simulation wrt the pin depth inside the cavity.

    Args:
        filepath (str): Path to the data

    Returns:
        None
    """
    file = pd.read_csv(filepath)
    n = file.shape[1]
    
    pin_lengths = file.iloc[:, 0].values
    freqs = file.iloc[:, 1:(n//2+1)].values
    Qs = file.iloc[:, (n//2+1):n].values
    
    plt.figure(figsize=(10,6))
    ax_freq = plt.subplot(2, 1, 1)
    ax_Q = plt.subplot(2, 1, 2)
    
    ax_freq.plot(-pin_lengths - 5, freqs*1e-9)
    ax_Q.plot(-pin_lengths - 5, Qs)
    
    ax_Q.set_xlabel("Pin Depth (mm)")
    ax_freq.set_ylabel("Frequency (GHz)")
    ax_Q.set_ylabel("Quality factor Q")
    ax_freq.grid()
    ax_Q.grid()
    
    plt.suptitle("Eigenmodes of the cavity with a pin")
    
    plt.tight_layout()
    plt.show()
```

```python
def plot_cavity_eigenmodes_eigenQ_ypin(filepath: str):
    """
    Plot eigenmodes and Q-factors from the simulation wrt the position of the pin inside the cavity.

    Args:
        filepath (str): Path to the data

    Returns:
        None
    """
    file = pd.read_csv(filepath)
    n = file.shape[1]
    
    pin_pos = file.iloc[:, 0].values
    freqs = file.iloc[:, 1:(n//2+1)].values
    Qs = file.iloc[:, (n//2+1):n].values
    
    plt.figure(figsize=(10,6))
    ax_freq = plt.subplot(3, 1, 1)
    ax_Q = plt.subplot(3, 1, 2)
    ax_Qzoom = plt.subplot(3, 1, 3)
    
    ax_freq.plot(pin_pos, freqs*1e-9)
    ax_Q.plot(pin_pos, Qs)
    ax_Qzoom.plot(pin_pos, Qs)
    ax_Qzoom.set_ylim(15000, 26000)
    
    ax_Qzoom.set_xlabel("Pin y-position (mm)")
    ax_freq.set_ylabel("Frequency (GHz)")
    ax_Q.set_ylabel("Quality factor Q")
    ax_Qzoom.set_ylabel("Quality factor Q")
    ax_freq.grid()
    ax_Q.grid()
    
    plt.suptitle("Eigenmodes of the cavity wrt the position of the pin")
    
    plt.tight_layout()
    plt.show()
```

```python
def plot_S11(filepath):
    file = pd.read_csv(filepath)
    
    figure, ax1 = plt.subplots(figsize=(10,6))
    
    ax1.plot(file.iloc[:, 0], file.iloc[:, 1], color='red')
    ax1.set_xlabel("Frequency (GHz)")
    ax1.set_ylabel("dB", color='red')
    ax1.tick_params(axis='y', labelcolor='red')
    
    ax2 = ax1.twinx()
    ax2.plot(file.iloc[:, 0],np.unwrap(file.iloc[:, 2], period=360), color='blue')
    ax2.set_ylabel("Phase (deg)", color='blue')
    ax2.tick_params(axis='y', labelcolor='blue')
    
    plt.title("$S_{11}$")

    plt.tight_layout()
    plt.show()
```

```python
def plot_S11_3D(filepath: str, dB_min: float = -5):
    file = pd.read_csv(filepath)

    freqs = file.iloc[:, 0].to_numpy()
    lengths = (file.iloc[:, 1]*10).to_numpy()
    dbs = file.iloc[:, 2].to_numpy()

    freqs_unique = np.sort(np.unique(freqs))
    lengths_unique = np.sort(np.unique(lengths))

    Z = dbs.reshape(len(lengths_unique), len(freqs_unique))

    plt.figure(figsize=(12, 6))

    im = plt.imshow(Z, aspect='auto', origin='lower',
                    extent=[freqs_unique.min(), freqs_unique.max(), lengths_unique.min(), lengths_unique.max()],
                   cmap='inferno', vmin=dB_min)

    cbar = plt.colorbar(im,  extend='min')
    cbar.set_label('dB')

    plt.xlabel('Frequency (GHz)')
    plt.ylabel('Cavity length (mm)')
    plt.title("$|S_{11}|^2$ (dB)")

    plt.tight_layout()
    plt.show()
```

```python
def plot_S11_phase_colorbar_old(filepath: str,
                           title: str = None):
    file = pd.read_csv(filepath)

    freq = file.iloc[:, 0].to_numpy()
    length = (file.iloc[:, 1]*10).to_numpy() # in mm
    phase = file.iloc[:, 2].to_numpy()

    freq_unique = np.sort(np.unique(freq))
    length_unique = np.sort(np.unique(length))

    Z_deg = phase.reshape(len(length_unique), len(freq_unique))
    Z_rad = np.deg2rad(Z_deg)
    Z_unwrap = np.unwrap(Z_rad, axis=1)
    Z_group_delay = -np.gradient(Z_unwrap, freq_unique, axis=1) / (2*np.pi)
    
    

    plt.figure(figsize=(12, 6))

    im = plt.imshow(Z_group_delay, aspect='auto', origin='lower',
                    extent=[freq_unique.min(), freq_unique.max(), length_unique.min(), length_unique.max()],
                   cmap='inferno',
                    norm = LogNorm(vmin=1, vmax=Z_group_delay.max()))
                   #norm=SymLogNorm(linthresh=1, linscale=1, vmin=Z_group_delay.min(), vmax=Z_group_delay.max(), base=10))

    cbar = plt.colorbar(im,  extend='min')
    cbar.set_label('Group delay (ns)')

    plt.xlabel('Frequency (GHz)')
    plt.ylabel('Cavity length (mm)')
    if title is None:
        plt.title("Group delay $S_{11}$")
    else:
        plt.title(title)

    plt.tight_layout()
    plt.show()
```

```python
def plot_S11_phase_colorbar(filepath: str,
                           title: str = None):
    file = pd.read_csv(filepath)

    freq = file.iloc[:, 1].to_numpy()
    length = (file.iloc[:, 0]).to_numpy() # in mm
    phase = file.iloc[:, 3].to_numpy()

    freq_unique = np.sort(np.unique(freq))
    length_unique = np.sort(np.unique(length))

    Z_deg = phase.reshape(len(length_unique), len(freq_unique))
    Z_rad = np.deg2rad(Z_deg)
    Z_unwrap = np.unwrap(Z_rad, axis=1)
    Z_group_delay = -np.gradient(Z_unwrap, freq_unique, axis=1) / (2*np.pi)
    
    

    plt.figure(figsize=(12, 6))

    im = plt.imshow(Z_group_delay, aspect='auto', origin='lower',
                    extent=[freq_unique.min(), freq_unique.max(), length_unique.min(), length_unique.max()],
                    cmap='inferno',
                    norm = LogNorm(vmin=1, vmax=Z_group_delay.max()))
                    #norm=SymLogNorm(linthresh=1, linscale=1, vmin=Z_group_delay.min(), vmax=Z_group_delay.max(), base=10))

    cbar = plt.colorbar(im,  extend='min')
    cbar.set_label('Group delay (ns)')

    plt.xlabel('Frequency (GHz)')
    plt.ylabel('Cavity length (mm)')
    if title is None:
        plt.title("Group delay $S_{11}$")
    else:
        plt.title(title)

    plt.tight_layout()
    plt.show()
```

```python
def plot_S11_comparison(filepath_list: list,
                       title: str = None):
    
    n = len(filepath_list)
    fig, axes = plt.subplots(nrows=n, ncols=1, figsize=(10, 4*n), sharex=True)
    
    for ax, filepath in zip(axes, filepath_list):
        file = pd.read_csv(filepath)
        
        freq = file.iloc[:, 0]
        dB = file.iloc[:, 1]
        phase = np.unwrap(file.iloc[:, 2], period=360)
        group_delay = -np.gradient(phase, freq) / (2*np.pi)

        ax.plot(freq, dB, color='red')
        ax.set_ylabel('Amplitude (dB)')
        ax.grid()

        ax2 = ax.twinx()
        ax2.plot(freq, group_delay, color='blue')
        ax2.set_ylabel('Group delay (ns)')

    axes[-1].set_xlabel('Frequency (GHz)')

    
    if title is None:
        plt.suptitle("Comparison $S_{11}$")
    else:
        plt.suptitle(title)
    plt.tight_layout()
    plt.show()
```
