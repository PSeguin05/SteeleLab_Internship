# Zaber control

This Jupyter Notebook gives different examples to use the ```Zaber_Linear_Stage``` class.

The two basic units are the second (s) and the millimeter (mm), such as
- Length is in [mm]
- Time is in [s]
- Velocity is in [mm/s]
- Acceleration is in [mm/s^2]

It is recommended to check the documentation **README.md** before any use of this class.

## Imports

````python
from Zaber_Linear_Stage import Zaber_Linear_Stage
````

### Import the motor

````python
motor = Zaber_Linear_Stage(port = "COM3",
                           label = "Linear Motor",
                           axis_number = 1,
                           auto_home = False,
                           default_velocity = 2,
                            default_acceleration = None,
                           auto_close = True,
                           wait_until_idle = True
                           ) 

motor.close() # Important to close the connection at the end of your script.
````

### Get parameters of the motor

````python
with Zaber_Linear_Stage(port = "COM3",
                        label = "Linear Motor",
                        axis_number = 1,
                        auto_home = True,
                        auto_close = True,
                        default_velocity = 2,
                        default_acceleration = None, 
                        wait_until_idle = True)

    motor.get_params()

# With a `with` statement, the connection will be automatically closed
# at the end of the block, even if an error occurs.
````

### Movements

````python
motor.move_absolute(position=0, velocity=1)
````

````python
motor.move_relative(delta_position=-5, velocity=0.5)
````

### Internal oscilloscope

````python
with Zaber_Linear_Stage(port = "COM3",
                           label = "Linear Motor",
                           axis_number = 1,
                           auto_home = False,
                           auto_close = True,
                           wait_until_idle = True
                           ) as motor:

    motor.scope_config(interval = 1e-2,
                       save_velocity = True)

    motor.scope_start()
    motor.move_absolute(position = 1, velocity=0.1)
    motor.scope_stop()

    result = motor.scope_save(filename = "Scope.csv")
    
````
